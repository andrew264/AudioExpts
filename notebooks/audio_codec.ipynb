```python
import sys
sys.path.append("..")

import torch

from model.audio_codec.model import AudioCodecConfig, AudioCodecModel
```

```python
model = AudioCodecModel(cfg = AudioCodecConfig())
device = 'cuda' if torch.cuda.is_available() else 'cpu'
model.to(device)
```

```python
weights = torch.load('../weights/audio_codec.ckpt', weights_only=True)
model.load_state_dict(weights, strict=False)
```

```python
weights = torch.load('../weights/trained_audio_codec.pt', weights_only=True)
model.load_state_dict(weights, strict=False)
```

```python
import librosa
import numpy as np
audio, _ = librosa.load('../data/test.mp3', sr=model.sample_rate)
audio_tensor = torch.from_numpy(np.array([audio, audio])).to(device)
```

```python
with torch.no_grad():
  encoded_tokens, encoded_len = model.encode(audio=audio_tensor)
```

```python
with torch.no_grad():
    reconstructed_audio, _ = model.decode(tokens=encoded_tokens,)
```
